```python
"""
Aqui está o passo a passo detalhado para instalar e configurar o Tesseract-OCR no Windows, para que seu código Python com 
pytesseract funcione corretamente:

_______________________________________________________________________________________________________________________________________________
Passo 1: Baixar o instalador do Tesseract
Acesse a página de instalação para Windows do Tesseract no repositório Mannheim (muito usado):
https://github.com/UB-Mannheim/tesseract/wiki

Nessa página, vá até a seção Installation e clique no link do instalador para Windows mais recente, por exemplo:
tesseract-ocr-w64-setup-v5.3.0.20221224.exe (o número pode variar conforme a versão).

Faça o download deste arquivo.
_______________________________________________________________________________________________________________________________________________
Passo 2: Instalar o Tesseract
Execute o instalador baixado (.exe).

Siga as instruções na tela do instalador:

Escolha o local de instalação (por exemplo, pode deixar o padrão em C:\Program Files\Tesseract-OCR).
Permita que o instalador conclua a instalação.
(Opcional) No instalador, você pode selecionar pacotes de idiomas adicionais, mas o pacote do idioma português geralmente já vem junto.
_______________________________________________________________________________________________________________________________________________
Passo 3: Encontrar o caminho do executável
Após a instalação, localize o arquivo tesseract.exe dentro da pasta de instalação, geralmente em:

C:\Program Files\Tesseract-OCR\tesseract.exe

_______________________________________________________________________________________________________________________________________________
Passo 4: Ajustar seu script Python para apontar o caminho do Tesseract
No seu script Python, importe o pytesseract e configure a variável tesseract_cmd, assim:

python


import pytesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
Coloque essa linha logo depois do import do pytesseract, antes de usar a função image_to_string.

_______________________________________________________________________________________________________________________________________________
Passo 5: Verificar se o Tesseract está funcionando no prompt de comando (opcional)
Abra o prompt de comando (cmd) e digite:

tesseract --version
Se retornar a versão do Tesseract, está instalado corretamente e no PATH.

Se não funcionar, siga normalmente a configuração do caminho no Python e ignore esse passo.

_______________________________________________________________________________________________________________________________________________
Passo 6: Executar seu script Python
Agora, seu script Python deve reconhecer o Tesseract e executar o OCR sem problemas.

"""
```

```python
import requests
from pdf2image import convert_from_path
import pytesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

import xml.etree.ElementTree as ET
import json

url = "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/09.23.12.59/doc/106-117.pdf"

pdf_filename = "documento.pdf"
html_filename = "documento.html"
xml_filename = "documento.xml"
json_filename = "documento.json"

poppler_path = r"C:\Users\profs\ICB_Trabalho_04_Análise_dos_Trabalhos_do _SBSR\Release-26.02.0-0\poppler-26.02.0\Library\bin"

def baixar_pdf(url, filename):
    print("Baixando PDF...")
    response = requests.get(url)
    response.raise_for_status()
    with open(filename, "wb") as f:
        f.write(response.content)
    print(f"PDF salvo como {filename}")

def converter_pdf_para_imagens(pdf_path):
    print("Convertendo PDF em imagens...")
    imagens = convert_from_path(pdf_path, poppler_path=poppler_path)
    print(f"{len(imagens)} páginas convertidas em imagens.")
    return imagens

def aplicar_ocr(imagens):
    print("Aplicando OCR nas imagens...")
    texto_completo = ""
    for i, img in enumerate(imagens):
        print(f"Lendo página {i + 1} com OCR...")
        texto_pagina = pytesseract.image_to_string(img, lang='por')
        texto_completo += texto_pagina + "\n"
    return texto_completo

def salvar_html(texto, filename):
    print(f"Salvando HTML em {filename}...")
    html_content = f"<html><body><pre>{texto}</pre></body></html>"
    with open(filename, "w", encoding="utf-8") as f:
        f.write(html_content)

def salvar_xml(texto, filename):
    print(f"Salvando XML em {filename}...")
    root = ET.Element("documento")
    pagina = ET.SubElement(root, "texto")
    pagina.text = texto
    tree = ET.ElementTree(root)
    tree.write(filename, encoding="utf-8", xml_declaration=True)

def salvar_json(texto, filename):
    print(f"Salvando JSON em {filename}...")
    dados = {"texto": texto}
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(dados, f, ensure_ascii=False, indent=4)

def main():
    baixar_pdf(url, pdf_filename)
    imagens = converter_pdf_para_imagens(pdf_filename)
    texto_extraido = aplicar_ocr(imagens)
    salvar_html(texto_extraido, html_filename)
    salvar_xml(texto_extraido, xml_filename)
    salvar_json(texto_extraido, json_filename)
    print("Processamento concluído!")

if __name__ == "__main__":
    main()
```

```python
import requests
from pdf2image import convert_from_path
import pytesseract
import xml.etree.ElementTree as ET
import json
import csv

url = "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/09.23.12.59/doc/106-117.pdf"

pdf_filename = "documento.pdf"
html_filename = "documento.html"
xml_filename = "documento.xml"
json_filename = "documento.json"
csv_filename = "documento.csv"

poppler_path = r"C:\Users\profs\ICB_Trabalho_04_Análise_dos_Trabalhos_do _SBSR\Release-26.02.0-0\poppler-26.02.0\Library\bin"

pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

def baixar_pdf(url, filename):
    print("Baixando PDF...")
    response = requests.get(url)
    response.raise_for_status()
    with open(filename, "wb") as f:
        f.write(response.content)
    print(f"PDF salvo como {filename}")

def converter_pdf_para_imagens(pdf_path):
    print("Convertendo PDF em imagens...")
    imagens = convert_from_path(pdf_path, poppler_path=poppler_path)
    print(f"{len(imagens)} páginas convertidas em imagens.")
    return imagens

def aplicar_ocr(imagens):
    print("Aplicando OCR nas imagens...")
    texto_completo = ""
    for i, img in enumerate(imagens):
        print(f"Lendo página {i + 1} com OCR...")
        texto_pagina = pytesseract.image_to_string(img, lang='por')
        texto_completo += texto_pagina + "\n"
    return texto_completo

def salvar_html(texto, filename):
    print(f"Salvando HTML em {filename}...")
    html_content = f"<html><body><pre>{texto}</pre></body></html>"
    with open(filename, "w", encoding="utf-8") as f:
        f.write(html_content)

def salvar_xml(texto, filename):
    print(f"Salvando XML em {filename}...")
    root = ET.Element("documento")
    pagina = ET.SubElement(root, "texto")
    pagina.text = texto
    tree = ET.ElementTree(root)
    tree.write(filename, encoding="utf-8", xml_declaration=True)

def salvar_json(texto, filename):
    print(f"Salvando JSON em {filename}...")
    dados = {"texto": texto}
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(dados, f, ensure_ascii=False, indent=4)

def salvar_csv(texto, filename):
    print(f"Salvando CSV em {filename}...")
    linhas = texto.strip().splitlines()
    with open(filename, mode='w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f, delimiter=';')
        for linha in linhas:
            writer.writerow([linha])

def main():
    baixar_pdf(url, pdf_filename)
    imagens = converter_pdf_para_imagens(pdf_filename)
    texto_extraido = aplicar_ocr(imagens)
    salvar_html(texto_extraido, html_filename)
    salvar_xml(texto_extraido, xml_filename)
    salvar_json(texto_extraido, json_filename)
    salvar_csv(texto_extraido, csv_filename)
    print("Processamento concluído!")

if __name__ == "__main__":
    main()
```
